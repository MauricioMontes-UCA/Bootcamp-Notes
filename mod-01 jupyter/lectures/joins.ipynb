```python
import pandas as pd

clientes = pd.DataFrame({
    "id_cliente": [1, 2, 3, 5],
    "nombre": ["Ana", "Luis", "Carlos", "Marta"],
    "departamento": ["SS", "SM", "CH", "SS"]
})

ventas = pd.DataFrame({
    "id_cliente": [1, 2, 3, 4],
    "producto": ["A", "B", "C", "D"],
    "cantidad": [10, 29, 8, 92]
})
```

```python
clientes.head(), ventas.head()
```

```python
# this is an inner join, basically
join_clientes_ventas = pd.merge(clientes, ventas, on="id_cliente", how="inner")
join_clientes_ventas
```

```python
# this is a left join, where clientes is shown completely
# but is there is not a register on the other dataframe, its filled with NaN's
pd.merge(clientes, ventas, on="id_cliente", how="left")
```

```python
# same with this one, a right join
pd.merge(clientes, ventas, on="id_cliente", how="right")
```

```python
# oh, yeah, and there are outer joins
pd.merge(clientes, ventas, on="id_cliente", how="outer")
```

```python
# retomando que: join_clientes_ventas = pd.merge(clientes, ventas, on="id_cliente", how="inner")
join_clientes_ventas[["nombre","producto"]]
```

Para mayor rigidez, es posible trabajar con una forma más parecida a la forma de las bases de datos, usando los parámetros `left_on`, `right_on`; para establecer la llave de una tabla y de la otra.

```python
pd.merge(clientes, ventas, left_on="id_cliente", right_on="id_cliente", how="inner")
```

```python
df_left = pd.merge(clientes, ventas, left_on="id_cliente", right_on="id_cliente", how="left")
df_left.dropna()
```
