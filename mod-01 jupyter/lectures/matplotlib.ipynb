```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Año": list(range(2010, 2021)),
    "El Salvador": [25.1, 25.3, 25.4, 25.5, 25.6, 25.8, 26.0, 26.1, 26.3, 26.4, 26.6],
    "Alemania": [9.5, 9.6, 9.7, 9.8, 9.9, 10.0, 10.2, 10.3, 10.4, 10.5, 10.6],
    "India": [24.0, 24.2, 24.3, 24.5, 24.7, 24.8, 25.0, 25.2, 25.3, 25.4, 25.6],
    "Canada": [2.0, 2.1, 2.3, 2.5, 2.6, 2.7, 2.9, 3.0, 3.2, 3.3, 3.5],
    "Brasil": [22.0, 22.1, 22.2, 22.3, 22.5, 22.6, 22.7, 22.9, 23.0, 23.2, 23.4],
    "Japón": [13.0, 13.1, 13.3, 13.4, 13.5, 13.6, 13.7, 13.8, 13.9, 14.0, 14.2],
}
```

```python
df = pd.DataFrame(data)

df.head()
```

```python
# Lienzo de trabajo, de 8" x 5"
plt.figure(figsize=(8,5))

# desde la columna 1 hasta la última columna
for col in df.columns[1:]:
    plt.plot(df['Año'], df[col])

plt.show()
```

La gráfica no es muy útil, dado que no tenemos etiquetas de gráfica, y muchas categorías que estamos comparando, por lo que puede ser abrumador y poco comprensible

```python
plt.figure(figsize=(8,5))
plt.plot(
    df['Año'],
    df['El Salvador'],
    label="El Salvador",
    marker='o',
    # color="green"
)
plt.title("Temperatura media anual (2010 - 2020)")
plt.xlabel("Año")
plt.ylabel("Temperatura promedio (°C)")
plt.legend()
plt.grid(True)

plt.xticks(df['Año'])

plt.show()
```

Ahora está mejor, aunque hay cosas que se pueden mejorar con las escalas. Por ejemplo, años que no aparecen o una escala medio extraña de las temperaturas.

Ahora, vamos a ver cómo metemos el resto de los países en la gráfica.

```python
plt.figure(figsize=(8,5))
plt.plot(
    df['Año'],
    df['El Salvador'],
    label="El Salvador",
    marker='o',
    # color="green"
)
plt.plot(
    df['Año'],
    df['Alemania'],
    label="Alemania",
    marker='o',
    # color="green"
)
plt.plot(
    df['Año'],
    df['Japón'],
    label="Japón",
    marker='o',
    # color="green"
)
plt.plot(
    df['Año'],
    df['Canada'],
    label="Canadá",
    marker='o',
    # color="green"
)
plt.plot(
    df['Año'],
    df['India'],
    label="India",
    marker='o',
    # color="green"
)
plt.plot(
    df['Año'],
    df['Brasil'],
    label="Brasil",
    marker='o',
    # color="green"
)
plt.xticks(df['Año'])

plt.title("Temperatura media anual (2010 - 2020)")
plt.xlabel("Año")
plt.ylabel("Temperatura promedio (°C)")
plt.legend(
    bbox_to_anchor=(1.05, 1), 
#    loc='upper left',
#    borderaxespad=0
)
plt.grid(True)
plt.show()
```

Ahora, tenemos un problema similar al de la primera gráfica; pero esto lo intentamos resolver con que la etiqueta que contiene las leyendas de las gráficas ocultan contenido.

```python
import numpy as np

np.random.seed(42)
temperaturas = np.random.normal(loc=15, scale=5, size=100)
temperaturas
```

```python
plt.figure(figsize=(8,5))
n, bins, patches = plt.hist(
    temperaturas,
    bins=10,
    color="skyblue",
    edgecolor="black"
)

plt.xticks(bins)
plt.grid(True, linestyle='--')
plt.show()
```
