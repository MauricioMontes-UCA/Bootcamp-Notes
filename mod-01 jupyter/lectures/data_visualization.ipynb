# Visualización de datos

Como dato, Seaborn permite la visualización de datos; sin embargo, son visualizaciones sencillas.

```python
import pandas as pd
import seaborn as sns
```

```python
data = {
    "Fruta": ["Manzanas", "Bananas", "Cerezas"],
    "Cantidad": [10, 15, 7]
}

df = pd.DataFrame(data)
```

```python
df
```

Tras definir un dataframe sencillo, haremos un gráfico de barras usando Seaborn:

```python
sns.barplot(data=df, x="Fruta", y="Cantidad")
```

Una opción más robusta, pues es más flexible es utilizar `matplotlib`.

```python
import matplotlib.pyplot as plt

plt.title("Cantidad de frutas")
plt.xlabel("Frutas")
plt.ylabel("Cantidades")
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots()

sns.barplot(data=df, x="Fruta", y="Cantidad", ax=ax)

ax.set_title("Cantidad de frutas")
ax.set_xlabel("Frutas")
ax.set_ylabel("Cantidades")

fig.tight_layout()
plt.show()
```
