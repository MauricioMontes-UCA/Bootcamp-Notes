# Limpieza de datos

Seaborn parece ser una herramienta para utilizar varios datasets.

```python
import pandas as pd
import seaborn as sns
```

Como se puede ver aquí, establecemos un dataframe llamando un dataset de Seaborn. En este caso estamos llamando al dataset "titanic".

```python
# Cargar el dataset
df = sns.load_dataset("titanic")

# Verificar que cargó.
df.head() # o... df.tail()
```

Ahora, queremos ver cómo está estructurada el dataset, y ver los tipos de datos de cada columna.

```python
df.info()
```

De un dataset limpio se espera que la cuenta de datos no nulos sea el mismo para todas las columnas, pero vemos que esto no se cumple:

Las columnas `age`, `embarked`, `embark_town` y `deck` no tienen la misma cantidad que el resto... hay que limpiar el dataframe.

Otro análisis del dataframe, más estadístico, se realiza con el método `describe()`.

```python
df.describe(include='all')
```

Primero, podemos ver que a la columna `age` se le realizó un análisis estadístico, esto se realizó sobre los datos que no son nulos. La media es de $29.7$ aproximadamente; y hay una desviación estándar algo alta, de $14.5$. Esto es porque también se puede ver los valores máximos y mínimos, que se alejan bastante de la media.

Por otro lado, la columna `sex` es una columna categórica, por lo que no hay un análisis numérico de esta columna. Sin embargo, hay una cuenta de cuántas categorías hay, la moda y la frecuencia de la moda.

Hay que tener cuidado de ver que los valores enteros son tratados como números reales.

```python
df.isnull()
```

Una cosa que hay que mencionar es que hay `None` y `NaN` no son los mismo; pues si bien `NaN` es considerado un valor numérico, `None` es un valor nulo o vacío.

Los motores de bases de datos no saben qué es `NaN`, así que hay que tener cuidado.

```python
df.isnull().sum()
```

Otro tipo de análisis que se puede hacer es el de observar si hay registros duplicados o no; que es otro problema de los datasets.

```python
df.duplicated().sum()
```

Entonces, de momento sabemos que hay columnas que poseen valores nulos y, para variar, el dataset tiene 107 registros duplicados. Divertido.

```python
df[df.duplicated()]
```

¿Qué podemos hacer con los datos nulos? Podemos eliminar el registro dependiendo del objetivo; pero en este caso vamos a reparar los datos sucios. Hay varios métodos que se pueden realizar para limpiar los datos y estos, nuevamente, dependen del objetivo.

Idealmente, hay que recuperar la información; rellenar con información que no altere mucho el significado del dataframe. Actualmente, para no afecta a la desviación estándar, los valores de edad nulos son rellenados con la edad promedio.

```python
df['age'] = df['age'].fillna(df['age'].mean())

df.describe()
```

Otro método es sacar el promedio de los cuartiles, y a cada cuartil rellenar con esa media.

```python
df.groupby('pclass')['age'].mean().transform(lambda x: x.fillna(x.mean()))

df.describe()
```

Una cosa que también se puede arreglar, que ya no es depuración de los datos, sino que normalización, es utilizar un valor equivalente para estandarizar el detalle de que la columna `sex` utiliza "male" y "female" para los datos.

```python
df['sex'] = df['sex'].replace({"male": "M", "female": "F"})

df
```

El caso de las columnas `deck` y `embark_town` es que no podemos conocer cuál era el valor que representan ahora los valores nulos. Una opción, dependiendo del tipo de análisis, es simplemente prescindir de la columna.

```python
df.drop(columns=["deck", "embark_town"], inplace = True)

# misma situacion con la columna embarked
df.drop(columns=["embarked"], inplace=True)
```

```python
df
```

Podemos ver una última cosa, y es ver que la columna de `alive` es de tipo objeto, a pesar de que parece ser que quiere ser un valor booleano. Lo mejor es estandarizarlo.

```python
df.info()
```

```python
df['alive'].unique()
```

```python
df['alive'] = df['alive'].replace({"no": False, "yes": True})

df.info()
```

Viendo el dataframe, a primera vista, vemos que las columnas `sex`, `who` y `adult_male` parecen ser que tratan de decir lo mismo; pero antes de deshacerse de las columnas, podemos observar mejor estas columnas.

```python
df[['sex', 'who', 'adult_male']].drop_duplicates()
```

```python
df['who'].unique()
```

Así, se puede ver que las columnas cumplen con una función que no es duplicada. Ahora, viendo que hay niños; hay un problema, y es que la edad que fue llenada en los datos nulos no es correcto, pues estamos asignando edades de 20 años a registros catalogados como niños.
