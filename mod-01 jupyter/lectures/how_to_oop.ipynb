# Programación Orientada a Objetos en Python

Yeah, at this point, I should know what is OOP, and what a class is. The thing is, how does it work with Python?

```python
import pandas as pd
```

```python
class myClass:
    def __init__(self):
        self.name = None
```

La clase `myClass` tiene un método inicializador, la función constructora. En esta función constructora establece que tiene un atributo `name`.

El caso de arriba tiene establecido por defecto el atributo "None".

```python
class MyClass:
    def __init__(self, name: str, age: int):
        self._name = name
        self._age = age

    def print_hello(self):
        print(f"Hola, soy {self.name} y tengo {self.age} años")

    @property
    def name(self):
        return self._name
    
    @name.setter
    def name(self, name: str):
        self._name = name

    @property
    def age(self):
        return self._age
    
    @age.setter
    def age(self, age: int):
        self._age = age
```

La clase de arriba es del estilo de Java hasta cierto punto.

```python
obj = MyClass("Mauricio", 20)
obj.print_hello()
```

Uno de los elementos importantes de la programación orientada a objetos es la herencia.

```python
class ChildClass(MyClass):
    def __init__(self, name: str, age: int, pronouns: str):
        super().__init__(name, age)
        self._pronouns = pronouns

    def show_pronouns(self):
        print(f"My pronouns are {self.pronouns}")

    @property
    def pronouns(self):
        return self._pronouns
    
    @pronouns.setter
    def gender(self, gender: str):
        self._pronouns = gender
```

```python
obj = ChildClass(
    name= "Mauricio",
    age= 20,
    pronouns= "He / Him"
)

obj.show_pronouns()
```
