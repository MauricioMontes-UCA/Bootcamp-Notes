## Sobre Pandas

Es una librería que permite trabajar con tablas o matrices de datos de una forma muy parecida a cómo se trabajaría una tabla de bases de datos o de excel, pero por medio de com

```python
import pandas as pd
```

En Pandas, una serie es un arreglo unidimensional de datos. El tipo de dato de la serie depende de su contenido.

```python
S = pd.Series([10, 20, 30, 40.1, 50])
print(S)
```

Por otro lado, cuando se tienen arreglos de dos dimensiones, este se le llama _data frame_. Los data frames pueden ser trabajados como objetos (del tipo de JavaScript), para así crear una tabla con encabezados.

El contenido de cada lista tiene que tener la misma cantidad de datos para que el _data frame_ funcione.

```python
datos = {
    'Nombre': ['Ana', 'Luis'],
    'Edad': [22, 35]
}

df = pd.DataFrame(datos)
print(df)
```

Además, los _data frames_ pueden exportarse como varios otros tipos de archivos. Es posible añadir más formatos para los archivos en cuestión; como por ejemplo el siguiente código exportará el data frame de arriba incluyendo una columna de índices.

```python
df.to_csv("ejemplo.csv")
```

Así como se pueden exportar los data frames, pueden importarse desde varios archivos.

```python
imported = pd.read_csv("ejemplo.csv")
print(imported)
```

Ahora, a la hora de importar un data frame, queremos verificar que la importación a sido exitosa. Si el data frame es muy grande, es mejor verificar solo con las primeras o últimas filas; pero no todo el data frame.

```python
# Imprime las primeras cinco filas del data frame
imported.head()
```

```python
# Imprime las últimas cinco filas del data frame
imported.tail()
```

Otra función particularmente útil es aquella que nos da la información del data frame, del tipo de datos de las columnas y cantidad de datos.

```python
df.info()
```

También se tiene la función para realizar una descripción estadística del data frame por columna.

```python
df.describe()
```

También es posible trabajar con una cantidad de columnas que se desean, no tiene que ser todo el data frame.

```python
filtered_df = df['Nombre']
print(filtered_df)
```

También es posible ver el tamaño del data frame, en las dos dimensiones. Esto es porque el data frame tiene un atributo, una tupla que empareja la cantidad de filas y columnas del data frame respectivamente.

```python
print(df.shape)
```

Entre otros atributos, están las columnas, que son tratadas como una estructura de datos. Por ejemplo:

```python
columns = list(df.columns)
print(columns)
```

Otra forma de verificar el tipo de datos que tienen cada columna del data frame es viendo al atributo `dtypes` del data frame.

```python
print(df.dtypes)
```

Una opción que nos da Pandas es la capacidad de ver el valor de una celda en específico. Esto se hace con el atributo `iloc`.

```python
print(int(df.iloc[0,1]))
```

Ahora, podemos verificar condicionales al data frame completo. Incluso, es posible obtener un data frame filtrado o discriminado respecto a la condicional dada.

```python
df[df['Edad'] < 30]
```

Pueden usarse varios condicionales, unidos con operadores lógicos como `and`o `or`. Dentro de Pandas, `and = &` y `or = |`. Por ejemplo:

`df[(df['Scores'] < 21) & (df['Hours'] < 5)]`

Nótese que cada condición tiene que estar entre paréntesis.

Una forma para depurar datos, verificar si hay valores nulos, es por medio del método `isnull()`, el cual debería devolver un data frame con los valores `false` si los valores no son nulos, y `true`si lo son. Para mejor vista, se agrega el método `sum()` para ver cuántos valores hay nulos de cada columna.

```python
print(df.isnull().sum())
```

Un método no recomendable, pero que existe, es `dropna()`, el cual borra todos los valores nulos

```python
df.dropna()
```

Por medio de Pandas, se pueden agregar columnas, y se pueden alterar los valores de columnas.

```python
df['Puntos'] = 5
df
```

```python
df['Puntos'] = 120
df
```

Se pueden renombrar columnas...

```python
df.rename(columns={'Puntos':'Puntaje'}, inplace=True)
df
```

Se pueden borrar columnas...

```python
df.drop(columns={'Puntaje'}, inplace=True)
df
```
