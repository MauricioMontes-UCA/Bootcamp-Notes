### Clustering
Es una técnica donde se agrupan datos que se parecen entre sí. Como por ejemplo, agrupar clientes por sus hábitos de compra. Para este tipo de modelado, se utiliza el algoritmo K-Means.

1. Elige un número K de grupos.
2. Se colocan K puntos aleatorios llamados centroides.
3. Se asignan los puntos al centroide más cercano.
4. Se recalculan los centroides.
5. Se repite hasta que no cambien más.

Este modelo, en resumen, toma una nube de puntos conocida y en base a su algoritmo clasifica los puntos en diferentes grupos a las que el centroide termina asignado. Cabe mencionar que los puntos no necesariamente son bidimensionales (relación entre dos variables); sino que puede usarse para agrupar datos con más variables relacionadas (multidimensional).

### DBSCAN
"Density-based spatial clustering of applications with noise". 

- La estrategia de este algoritmo es establecer una mínima cantidad de puntos que se encuentran a una distancia $\varepsilon$ de un punto $p$ para establecer a este punto como un punto núcleo.

- Si un punto $q$ solo cumple la condición de estar a una distancia $\varepsilon$ de un punto $p$, este punto es considerado como directamente alcanzable.

- Por último, si un punto no cumple con ninguna de las condiciones, este punto es considerado ruido.
